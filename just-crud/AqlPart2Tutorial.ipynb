## Filter

To find documents that fulfill certain criteria more complex than key equality, there is the FILTER operation in AQL, which enables us to formulate arbitrary conditions for documents to match. We actually have used a filter condition before to find `
Ned`:

```python
find_ned_query = """
FOR c IN Characters
    FILTER c.name == "Ned"
    RETURN c
"""
# python-arango
query_result = aql.execute(find_ned_query)

# pyArango
# query_result = db.AQLQuery(find_ned_query, rawResults=True)
for doc in  query_result:
    print(doc)
    print()
```

```python
find_ned_query = """
FOR c IN Characters
    FILTER c.surname == "Stark"
    RETURN c
"""
# python-arango
query_result = aql.execute(find_ned_query)

# pyArango
# query_result = db.AQLQuery(find_ned_query, rawResults=True)
for doc in  query_result:
    print(doc)
    print()
```

## Range conditions
Strict equality is one possible condition we can state. There are plenty of other conditions we can formulate however. 
For example, we could ask for all young characters:

```python
find_adults_query = """
FOR c IN Characters
    FILTER c.age >= 13
    RETURN {name: c.name, age: c.age }
"""
# python-arango
query_result = aql.execute(find_adults_query)

# pyArango
# query_result = db.AQLQuery(find_adults_query, rawResults=True)
for doc in  query_result:
    print(doc)
    print()
```

For more about **Match document** query, [here](https://docs.arango.ai/arangodb/stable/get-started/start-using-aql/filter/) 

## Sort by multiple attributes
Assume we want to sort by surname. Many of the characters share a surname. The result order among characters with the same surname is undefined. We can first sort by surname, then name to determine the order:

```python
all_characters_names = """
FOR c IN Characters
    FILTER c.surname
    SORT c.surname, c.name
    LIMIT 10
    RETURN {
        surname: c.surname,
        name: c.name
    }
"""

query_result = aql.execute(all_characters_names)

for doc in  query_result:
    print(doc)
    print()
```

```python
sample_chars_query = """
FOR c IN Characters
    SORT c.name
    LIMIT 2, 5
    RETURN c.name
"""
# python-arango
query_result = aql.execute(sample_chars_query)

# pyArango
# query_result = db.AQLQuery(sample_chars_query, rawResults=True)
for doc in  query_result:
    print(doc)
    print() 
```

For more about **Sort and Limit**, [here](https://docs.arango.ai/arangodb/stable/get-started/start-using-aql/sort-limit/)
